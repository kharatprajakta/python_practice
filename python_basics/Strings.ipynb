```python
#
n=int(input('enter n'))
result=0
fact=1
for i in range(1,n+1):
    fact=fact*i
    result=result+i/fact
print(result)    
```

```python
rows=int(input('enter number of rows'))

for i in range(1,rows+1):
    for j in range(1,i+1):
        print('*',end='')
    print() 
```

```python
rows=int(input("enter number of rows"))

for i in range(1,rows+1):
    for j in range(1,i+1):
        print(j,end="")
    for k in range(i-1,0,-1):
         print(k,end='')
    print()    
    
```

```python
#Capitalize/Title/Upper/Lower/Swapcase
s="hello world"
print(s.capitalize())
print(s)
print(s.title())
print(s.upper())
print(s.lower())
print('heLlo WoRlD'.swapcase())
```

```python
#Count/Find/Index
print('my name is prajakta'.count('a'))
print('my name is prajakta'.find('prajakta'))
print('my name is prajakta'.find('c'))
print('my name is prajakta'.index('j'))
```

```python
# endswith/startwith
print('my name is prajakta'.endswith('sho'))
print('my name is prajakta'.startswith('my'))
```
