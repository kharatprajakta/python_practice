```python
# Adding items to a list

# append
L=[1,2,3,4,5]
L.append(6)
print(L)
L.append([11,12]) # in this command you add the another list
print(L)

# extend
L1=[1,2,3,4,5]
L1.extend([6,7,8]) # you can extend the list using the extend command
print(L1)
L2=[1,2,3]
L2.extend('delhi')
print(L2)

# insert
L3=[1,2,3,4,5]
L3.insert(1,100) # you can add only ones
print(L3)
```

```python
# editing items in a list
# you cannot change in the string but in case of list we can edit the list
L=[1,2,3,4,5]
# editing with indexing
L[-1]=500
print(L)

# editing with slicing
L[0:4]=[200,300,400]
print(L)
```

```python
# Deleting items from a list

#del
L=[1,2,3,4,5]
print(L)
#indexing
del L[-1]
print(L)

# slicing
L1=[1,2,3,4,5]
del L1[1:3]
print(L1)

# remove
L=[1,2,3,4,5]
L.remove(5)
print(L)

# remove
L=[1,2,3,4,5]
L.remove(2)
print(L)

# pop
L=[1,2,3,4,5]
L.pop(0) # default pop will delete the last entry in the list
print(L)

#
L=[1,2,3,4,5]
L.clear()
print(L)
```

```python
# List Comprehension
# newlist =[expression for item in iterable if condition == True]
# Add 1 to 10 numbers to a list
L=[]

for i in range(1,11):
    L.append(i)
print(L)
```

```python
# scalar multiplication on a vector
v=[2,3,4]
s=-3
x=[]
for i in v:
    x.append(i*s)
print(x)    
```

```python
# by list comprehension
v=[2,3,4]
s=-3
x=[s*i for i in v]
print(x)
```

```python
# Add squares
L=[1,2,3,4,5]
L1=[]
for i in L:
    L1.append(i**2)
print(L1)    

# Add Squares by list comprehension
L=[1,2,3,4,5]
[i**2 for i in L]
```

```python
# Nested if with List Comprehension
basket=['apple','guava','cherry','banana']
my_fruits=['apple','kiwi','grapes','banana']

# add new list from my_fruits and items if the fruit exists in basket and also starts with 'a'
[fruit for fruit in my_fruits if fruit in basket if fruit.startswith('a')]
```

```python
# print a (3,3) matrix using list comprehension -> Nested List comprehension
[[i*j for i in range(1,4)] for j in range(1,4)]
```

```python
# by using for loop
result=[]

for j in range(1,4):
    row=[]
    for i in range(1,4):
        row.append(i*j)
    result.append(row)
print(result)    
```

```python
# cartesian  products -> List Comprehension on 2 lists together
L1=[1,2,3,4]
L2=[5,6,7,8]
[i*j for i in L1 for j in L2]


# using for loop
L1=[1,2,3,4]
L2=[5,6,7,8]
result=[]
for i in L1:
    for j in L2:
        result.append(i*j)
print(result)        

```

```python
# 2 ways to traverse a list
# itemwise
L=[1,2,3,4]
for i in L:
    print(i)
# indexwise
L1=[1,2,3,4]
for i in range(0,len(L)):
    print(L[i])
```
