```python
print(scores.shape)
print(scores.ndim)
print(scores.size)
print(scores.dtype)
```

```python
print(scores)
ab=scores[0]
print(ab)
print(scores[1,1])
print(scores[:,0])
print(scores[0:4])
print(scores[:,[0,2]])
```

```python
print(scores.mean())
print(scores.mean(axis=0)) # axis=0 column
print(scores.mean(axis=1)) # axis=1 row
print(scores.max(axis=0))
print(scores.min(axis=0))
print(scores.sum(axis=1))
```

```python
med=np.median(scores,axis=0)
sd=np.std(scores,axis=0)
ag=np.argmax(scores[:,0])
print(med)
print(sd)
print(ag)
```

```python
print(scores+5)
print(scores*1.1)
print(scores/100)
```

```python
centered=scores-scores.mean(axis=0)
print(centered)
```

```python
normalized=(centered)/(scores.max(axis=0)-scores.min(axis=0))
print(normalized)
```

```python
# Boolean masking and filtering
print(scores[:,0]>70)
print(scores[scores[:,0]>70])
```

```python
scores[(scores[:,0]>70)&(scores[:,1]>80)&(scores[:,2]>65)]
```

```python
sc=np.where(scores>=60,"pass","fail") # lbel every score
print(sc)
cs=np.sum(scores<60)# how many scores are below 50
print(cs)
```
